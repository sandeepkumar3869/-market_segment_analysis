```python
import os
import zipfile

zip_name = "VC-Shortages-storemapsmanager-dev.zip"  # Change this to your actual zip filename

if not os.path.exists(zip_name):
    raise FileNotFoundError(f"File not found: {zip_name}")

with zipfile.ZipFile(zip_name, "r") as z:
    z.extractall(".")  # Extracts into the current notebook folder

print(f"Extracted '{zip_name}' to '{os.getcwd()}'")
```
